### データ取得の件


```python
# よく使うやつらインストール
# import pandas as pd
# import numpy as np
# import matplotlib.pyplot as plt 

# 試し
import genesis as gs
# gs.init(backend=gs.gpu)

# 設定・環境等の作り方試す
# Notebook LMの検索で一旦Hello world

scene = gs.Scene(
    sim_options=gs.options.SimOptions(dt=0.01,),# dt: 1回の物理ステップでどのくらい時間を進めるか. ここが大きいとぶつ切りになる
    show_viewer=False) 
# シーンに登場させたいOBJを追加する
# 関数で追加をしつつ, 追加されたOBJがinstanceになるのか
plane = scene.add_entity(gs.morphs.Plane()) 
robot = scene.add_entity(gs.morphs.URDF(file="urdf/go2/urdf/go2.urdf", pos=(0,0,10)))



```

```python
robot.get_contacts()
```

```python
robot.detect_collision
```

```python
print("[現状の話]")
print("関節パラメータ数:", robot.n_dofs)
print("関節の数:", robot.n_joints)
print("関節リスト:", robot.joints)



# 物理ステップを実行?
scene.build()

data_results = {
    "vel": [],
    "pos": [],
    "collision": [],
    "dofs_position": [],
    "dofs_velocity": [],
}

for i in range(500):
    data_results["pos"].append(robot.get_pos())

    data_results["vel"].append(robot.get_vel())
    data_results["collision"].append(robot.detect_collision())
    data_results["dofs_position"].append(robot.get_dofs_position())
    data_results["dofs_velocity"].append(robot.get_dofs_velocity())
    scene.step()
```
